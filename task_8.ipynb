**1. Encapsulation**
 * Create a class Person with private attributes __name and __age.
 * Use getter and setter methods to access and modify these values.


```python
class person:
    def __init__(self,name,age):     # encapsulation means hiding the details of an obj and exposing only what is necessary.
      self.__name = name
      self.__age = age

    def set_name(self,name):
      self.__name = name
    def get_name(self):
      return self.__name
    def set_age(self,age):
      self.__age = age
    def get_age(self):
      return self.__age # set lay value change garxa ,
                         # get lay value return garxa
p = person('anita',21)
p.set_name('ani')
p.set_age(20)
print(p.get_name())
print(p.get_age())
```

**2. Polymorphism**
 * Create a function show_details(obj) that works with different classes.
 * Define the same method name (details()) in two or more classes.
 * Call show_details() with different objects to show polymorphism.

```python
class student:
  def details(self):
    print("i an a student")
class teacher:
  def details(self):
    print("i an a teacher")
def show_details(obj):
  obj.details()
s=student()
t=teacher()
show_details(s)
show_details(t)

```

**4. Multiple Inheritance**

        - Create a class Performer that inherits from both.

**5. Hierarchical Inheritance**    
 * Create a base class Appliance.
    - Create two child classes: WashingMachine and Microwave.
    - WashingMachine should have a method wash(), and Microwave should have a method heat().
    - Create objects of both and call their respective methods.

```python
class applicance:            # 1 parent, multiple child
  def show(self):
    print("applicance :")
class washingmachine(applicance):
  def wash(self):
    print("washingmachine used to wash cloth")
class microwave(applicance):
  def heat(self):
    print("microwave used to heat food")
w=washingmachine()
m=microwave()
w.show()
w.wash()
m.show()
m.heat()
```

**7. Use Encapsulation to Protect Age**
  * Create a class Person with private attribute __age.
      - Use set_age() to set the age.
      - If age is less than 0, print "Invalid age".
      - Use get_age() to return the age.

```python
class person:
  def __init__(self,age):
    self.__age=0
  def set_age(self,age):
    if age>0:
      self.__age=age
    else:
      print("invalid age")
  def get_age(self):
    return self.__age
p=person(20)
p.set_age(10)
print(p.get_age())
```

**8. Company Structure**
  * Create class Company with method company_name().
      - Create class Department that inherits Company and adds department_name().
       - Create class Employee that inherits Department and adds employee_name().
       - Create an Employee object and call all the methods.

```python
class company:
  def company_name(self):
    print("company name:mindrise")
class department(company):
  def department_name(self):
    print("department name:data science")
class employee(department):
  def employee_name(self):
    print("employee name:unii")
e=employee()
e.company_name()
e.department_name()
e.employee_name()
```

**11. Add 10 to All ListElement**s
  * Given a list [1, 2, 3, 4, 5],
      - Use map() and lambda to return a new list where 10 is added to each number.

```python
num=[1,2,3,4,5] # map()-applies a function to all elements in the list or tuple
result=map(lambda x : x+10 , num)
print(list(result))
```

**14.Create a Word by Joining Letters**
  * Given a list of letters ['P', 'Y', 'T', 'H', 'O', 'N'],
      - Use reduce() and lambda to join them into a single word: "PYTHON".

```python
from functools import reduce # reduce(): imported from functool.
c=['p','y','t','h','o','n']
result=reduce(lambda x,y : x+y,c)
print(result)
```
